```python
from baby_pytorch import Value

import random
```

```python
num_samples = 5000
```

```python
def get_labels(x1, x2, r):
    return [
        -1 if (x1[i].data ** 2) + (x2[i].data ** 2) < (r ** 2) else 1
            for i in range(0, len(x1))
    ]
```

```python
def make_data(x1, x2):
    return [
    [x_1, x_2] for x_1, x_2 in zip(x1, x2)
]    
```

```python
x1 = [Value(random.uniform(-1, 1)) for _ in range(num_samples)]
x2 = [Value(random.uniform(-1, 1)) for _ in range(num_samples)]

r = 0.5

labels = get_labels(x1, x2, r)
```

```python
import matplotlib.pyplot as plt

def plot_data(x1, x2, labels):
    plt.scatter([v.data for v in x1], [v.data for v in x2],
                c=labels, cmap='bwr')
    plt.gca().set_aspect('equal')
    plt.show()
```

```python
plot_data(x1, x2, labels)
```

```python
from baby_pytorch.nn import MLP
from baby_pytorch.activation_functions import relu

from baby_pytorch.optim import SGD
from baby_pytorch.loss import MSE

num_epochs = 20
batch_size = 10

mlp = MLP(2, [16, 8], 1, relu)

optimizer = SGD(mlp.parameters(), lr=0.1)

inputs = make_data(x1, x2)

for epoch in range(num_epochs):
    for idx in range(0, num_samples, batch_size):
        batch = inputs[idx: idx + batch_size]
        optimizer.zero_grad()
        preds = [mlp(x)[0] for x in batch]

        loss = MSE(preds, labels[idx:idx + batch_size])
        loss.backward()

        batch_number = idx // batch_size
        if batch_number % 10 == 0:
            print(f"Epoch {epoch}, batch {idx // batch_size}: Loss: {loss.data}")
            
        optimizer.step()
```

```python
def get_prediction(logit):
    return -1 if logit.data < 0 else 1    
```

```python
def calculate_accuracy(preds, labels):
    correct = sum([1 if pred == label else 0 for pred, label in zip(preds, labels)])
    return 100.0 * correct / len(preds)
```

```python
train_preds = [get_prediction(mlp(x)[0]) for x in inputs]
print(f"Train accuracy: {calculate_accuracy(train_preds, labels):2f}%")
```

```python
num_test_samples = 1000
test_x1 = [Value(random.uniform(-1, 1)) for _ in range(num_test_samples)]
test_x2 = [Value(random.uniform(-1, 1)) for _ in range(num_test_samples)]

test_labels = get_labels(test_x1, test_x2, r)
plot_data(test_x1, test_x2, test_labels)
```

```python
def plot_errors(x1, x2, labels, preds):
    xs = [v.data for v in x1]
    ys = [v.data for v in x2]
    plt.scatter(xs, ys, c=labels, cmap='bwr', edgecolors='k', s=30)

    wrong = [i for i, (p, l) in enumerate(zip(preds, labels)) if p != l]
    plt.scatter([xs[i] for i in wrong], [ys[i] for i in wrong],
                c='lime', edgecolors='k', s=60, zorder=3)

    plt.gca().set_aspect('equal')
    plt.show()
```

```python
test_inputs = make_data(test_x1, test_x2)
test_preds = [get_prediction(mlp(x)[0]) for x in test_inputs]

accuracy = calculate_accuracy(test_preds, test_labels)
print(f"Accuracy: {accuracy:2f}%")
plot_errors(test_x1, test_x2, test_labels, test_preds)
```

```python
import numpy as np

# grid over the plane
gx = np.linspace(-1, 1, 50)
GX, GY = np.meshgrid(gx, gx)

# forward pass at every grid point
Z = np.array([
    mlp([Value(a, requires_grad=False), Value(b, requires_grad=False)])[0].data
    for a, b in zip(GX.ravel(), GY.ravel())
]).reshape(GX.shape)

# two bands split at zero -- the seam is the boundary
plt.contourf(GX, GY, Z, levels=[-1e9, 0, 1e9], colors=['#cfd8ff', '#ffd0d0'])

# your data on top
plt.scatter([v.data for v in x1], [v.data for v in x2],
            c=labels, cmap='bwr', edgecolors='k', s=30)

plt.gca().set_aspect('equal')

th = np.linspace(0, 2*np.pi, 200)
plt.plot(0.5*np.cos(th), 0.5*np.sin(th), 'k--')

plt.show()
```
